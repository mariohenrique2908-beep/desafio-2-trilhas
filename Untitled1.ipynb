```python
import pandas as pd

# 1. Lendo o arquivo original no caminho do Google Drive
df = pd.read_csv('/content/drive/MyDrive/ecommerce_sales_data_original.csv')

# 2. Convertendo a coluna de data
df['Order Date'] = pd.to_datetime(df['Order Date'])

# 3. Criando as novas colunas
df['Year'] = df['Order Date'].dt.year

# Converter o mês para string em português (opcional), mas vamos manter a lógica atual de name
df['Month'] = df['Order Date'].dt.month_name()
df['Day_of_Week'] = df['Order Date'].dt.day_name()
df['Quarter'] = df['Order Date'].dt.quarter
df['Unit_Price'] = df['Sales'] / df['Quantity']
df['Profit_Margin_%'] = (df['Profit'] / df['Sales']) * 100

# 4. Arredondando os valores
df['Unit_Price'] = df['Unit_Price'].round(2)
df['Profit_Margin_%'] = df['Profit_Margin_%'].round(2)

# 5. Salvando o novo arquivo tratado
df.to_csv('ecommerce_sales_data_tratado.csv', index=False)

print("Sucesso! O arquivo tratado foi gerado.")
```
